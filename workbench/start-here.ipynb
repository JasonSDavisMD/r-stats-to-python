# statsbench: start here

**statsbench** helps you find the right Python call for a statistics or machine-learning task. It covers the whole scientific Python stack (NumPy, SciPy, SymPy, pandas, statsmodels, scikit-learn, plus optional PyTorch, PyMC, XGBoost and more), with R equivalents, runnable examples and caveats.

Run each cell with **Shift+Enter**. On Binder nothing is saved; for your own work, see `docs/jupyterlab-guide.md`.

````python
from statsbench.finder import find

for hit in find("R wilcox.test", limit=3):
    print(f"{hit.entry.id:<28} [{hit.entry.kind}] {hit.entry.call}")
````

Each hit carries the import line, the exact call, caveats (especially where Python defaults differ from R), the official docs link, and a runnable example:

````python
entry = find("inverse logit")[0].entry
print(entry.import_line)
print(entry.docs)
print()
print(entry.example)
````

````python
# Run the example right here: the catalog examples are self-contained.
exec(entry.example)
````

## Same search from a terminal

**File -> New -> Terminal**, then:

```bash

stats kinds         # fit vs evaluate vs optimize vs solve
```
